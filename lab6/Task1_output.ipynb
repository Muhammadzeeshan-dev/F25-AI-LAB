## Part 1: Tokenization

This part compares whitespace splitting and a Unicode word-like regex on six examples.

```python
import re
```

```python


texts = [
    "Hello, how are you?",
    "I am not  attend today's class.",
    "My email is zeeshan@superior.com",
    "My course code is CS4944",
    "",
    "یہ لیب 6 کا میرا ٹاسک ہے۔"
]

print("TOKENIZATION TEST")

for text in texts:
    words = text.split()
    regex_words = re.findall(r"\b\w+(?:['’]\w+)*\b", text)

    print("\nOriginal text:", repr(text))
    print("Whitespace tokens:", words)
    print("Whitespace count:", len(words))
    print("Regex tokens:", regex_words)
    print("Regex count:", len(regex_words))

print("\nMODEL TOKENIZER TEST")

try:
    from transformers import AutoTokenizer

    tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

    examples = [
        "AOA, how are you dooing today?",
        "I am learning artificial intelligence."
    ]

    for text in examples:
        tokens = tokenizer.tokenize(text)
        print("\nText:", text)
        print("Model tokens:", tokens)
        print("Model count:", len(tokens))

except ImportError:
    print("Transformers library is not installed.")
    print("Model tokenizer test skipped.")
except Exception:
    print("Model tokenizer is not available.")
    print("Model tokenizer test skipped.")

print("\nOBSERVATIONS")
print("1. Whitespace keeps punctuation attached, while regex removes it.")
print("2. Whitespace keeps an email as one token, while regex splits it.")
print("3. Changing or removing Unicode characters may affect text meaning.")
print("Regex is not a complete multilingual tokenizer.")
```

## Part 2: FAQ Chatbot

This chatbot uses TF-IDF to compare a user's question with two FAQ questions. It gives an answer for an exact question, suggests a close question, and handles empty or unsupported input.

### Run the chatbot


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

questions = [
    "Why did I choose software engineering?",
    "What am I trying to build these days?"
]

answers = [
    "I chose software engineering because I like making websites and learning how apps work.",
    "These days I am working on LoanoraTools. It is my website for simple loan and finance calculations."
]

sources = [
    "My personal answer",
    "My project work"
]

vectorizer = TfidfVectorizer()
faq_vectors = vectorizer.fit_transform(questions)

print("My FAQ Chatbot")
print("Type exit to stop.")

while True:
    user = input("\nYou: ").strip()

    if user.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    if user == "":
        print("Chatbot: Please enter a question.")
        continue

    if user.lower() == questions[0].lower():
        print("Answer:", answers[0])
        print("Source:", sources[0])

    elif user.lower() == questions[1].lower():
        print("Answer:", answers[1])
        print("Source:", sources[1])

    else:
        user_vector = vectorizer.transform([user])
        scores = cosine_similarity(user_vector, faq_vectors)[0]
        best = scores.argmax()

        if scores[best] > 0.15:
            print("Chatbot: Maybe you are asking:", questions[best])
        else:
            print("Chatbot: Sorry, I don't know this one.")
```
